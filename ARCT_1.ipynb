# ARCT — historical-data simulation walkthrough

Install the dependencies in [README.md](README.md), then restart the kernel and run all cells. This notebook uses the same tested functions as the CLI. It reads `Resources csv/ARCT_data.csv` and needs no API keys or online data.

These are educational scenarios fitted to a historical snapshot ending in 2023. The input's split/dividend adjustment policy is unverified. The figures are not current quotes or investment recommendations.

```python
import matplotlib.pyplot as plt
from IPython.display import display

from risk_reward.analysis import analyze_portfolio
from risk_reward.data import load_prices

TICKER = "ARCT"
prices = load_prices([TICKER])
print(
    f"{TICKER}: {len(prices):,} observations, {prices.index[0].date()} to {prices.index[-1].date()}"
)
display(prices.head())
```

## Compare horizons

The simulator fits daily log-return mean and variance to the historical prices. It uses 252 simulated trading steps per year, 500 paths, and a fixed seed. The same seed means longer horizons share the start of shorter paths.

`mean_terminal_value` is the mean of all simulated endings, not the midpoint of two quantiles. `p025` and `p975` mark the central 95% of the simulated distribution; they are not a guarantee about future prices. Longer horizons compound the assumptions.

```python
result = analyze_portfolio(
    prices, years=(1, 3, 5, 8, 10, 15), simulations=500, seed=42, initial_investment=2500
)
display(result.summary.round(4))
```

## Inspect the first horizon

Only the first 100 of the 500 one-year paths are plotted for readability. The distribution and summary use every path.

```python
result.preview.plot_simulation()
plt.show()
result.preview.plot_distribution()
plt.show()
```

## Explain the result

Compare the mean and median, the width of the scenario range, and the fraction of simulated paths ending below their starting value. These describe the fitted model, not verified probabilities for an investment.
